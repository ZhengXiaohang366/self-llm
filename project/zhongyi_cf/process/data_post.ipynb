```python
import json
import re

version = '0308'

# 方法一：使用 with 语句自动管理文件的打开和关闭
file_path = f'../dataset/res/tmp_res1{version}.json'
with open(file_path, 'r', encoding='utf-8') as file:
    # 读取并解析 JSON 数据
    test_data1 = json.load(file)

file_path = f'../dataset/res/tmp_res2{version}.json'
with open(file_path, 'r', encoding='utf-8') as file:
    # 读取并解析 JSON 数据
    test_data2 = json.load(file)

# file_path = f'../dataset/res/tmp_res3{version}.json'
file_path = f'../dataset/res/tmp_res3.json'
with open(file_path, 'r', encoding='utf-8') as file:
    # 读取并解析 JSON 数据
    test_data3 = json.load(file)

res_path = f'../dataset/res/final_res{version}.json'
```

```python
id_list = [test_data1[i]['ID']  for i in range(len(test_data1))]
```

```python
for i in range(200):
    original_str = test_data3[i]['output']
    items = [f"'{item.strip()}'" for item in original_str.split(',')]
    target_str = f"[{', '.join(items)}]"
    test_data3[i]['output'] = target_str
```

```python
test_data1_prep = {}
test_data2_prep = {}
test_data3_prep = {}
for i in range(200):
    test_data1_prep[test_data1[i]['ID']] = test_data1[i]['output']
    test_data2_prep[test_data2[i]['ID']] = test_data2[i]['output']
    test_data3_prep[test_data3[i]['ID']] = test_data3[i]['output']
```

```python
final_res = []
for i in id_list:
    tmp_dic = {}
    tmp_dic['ID'] = i
    tmp_dic['子任务2'] = test_data3_prep[i]
    tmp_dic['子任务1'] = [test_data2_prep[i],test_data1_prep[i]]
    final_res.append(tmp_dic)
```

```python
final_res
```

```python
with open(res_path, 'w', encoding='utf-8') as file:
    # 将 list 转换为 JSON 格式并写入文件
    json.dump(final_res, file, ensure_ascii=False, indent=4)
```
